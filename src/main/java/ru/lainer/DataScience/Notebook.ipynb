```kotlin
%use lets-plot

@file:DependsOn("com.opencsv:opencsv:5.7.1")
@file:DependsOn("org.jetbrains.lets-plot:lets-plot-kotlin-jvm:4.10.0")
//@file:DependsOn("org.jetbrains.lets-plot:lets-plot-kotlin-jvm:4.4.1")
//@file:DependsOn("org.jetbrains.lets-plot:lets-plot-common:4.5.0")
//@file:DependsOn("org.jetbrains.lets-plot:lets-plot-batik:4.5.0")

import com.opencsv.CSVReader
import java.io.FileReader
import java.time.LocalDateTime
import java.time.format.DateTimeFormatter
import org.jetbrains.letsPlot.*
import org.jetbrains.letsPlot.geom.*
import org.jetbrains.letsPlot.label.*
import org.jetbrains.letsPlot.themes.elementText
import org.jetbrains.letsPlot.themes.theme

val pathToFile: String = "/home/source/srcCode/DSAGitHub004/src/main/java/ru/lainer/DataScience/pressure.csv"
val fileReader = FileReader(pathToFile)
val csvReader = CSVReader(fileReader)
val resultOfReadCSV = csvReader.readAll()
csvReader.close()

val xAxis = mutableListOf<String>() // Список дат
val yAxis = mutableListOf<Int>()    // Список значений давления
for(current in 1 until resultOfReadCSV.size) {
    xAxis.add(resultOfReadCSV[current][0])
    yAxis.add(resultOfReadCSV[current][1].toInt()) //1-Верхнее давление, 2-Нижнее давление
}

// Создаём DataFrame-подобную структуру для lets-plot
val plotData = mapOf(
    "date" to xAxis,
    "pressure" to yAxis
)

// Строим график
val plot = letsPlot(plotData) +
        geomLine(color = "blue", size = 1.0) { x = "date"; y = "pressure" } +
        geomPoint(color = "red", size = 4.0) { x = "date"; y = "pressure" } +
        ggtitle("Динамика верхнего давления") +
        xlab("Дата и время") +
        ylab("Верхнее давление (мм рт. ст.)") +
        theme(axisTextX = elementText(angle = 45, hjust = 1))  // Наклон подписей дат

plot
```

```kotlin
// Можно добавить расчет корреляции

@file:DependsOn("com.opencsv:opencsv:5.7.1")
@file:DependsOn("org.nield:kotlin-statistics:1.2.1") //от 2018 года

import com.opencsv.CSVReader
import java.io.FileReader
import org.nield.kotlinstatistics.descriptiveStatistics
import org.nield.kotlinstatistics.median
import org.nield.kotlinstatistics.mode
import org.nield.kotlinstatistics.percentile
import org.nield.kotlinstatistics.percentileBy

val pathToFile: String = "/home/source/srcCode/DSAGitHub004/src/main/java/ru/lainer/DataScience/pressure.csv"
val fileReader = FileReader(pathToFile)
val csvReader = CSVReader(fileReader)
val resultOfReadCSV = csvReader.readAll()
csvReader.close()

val upperPressure = mutableListOf<Double>() // Список значений верхнего давления
val lowerPressure = mutableListOf<Double>() // Список значений нижнего давления

for(current in 1 until resultOfReadCSV.size) {
    upperPressure.add(resultOfReadCSV[current][1].toDouble()) //1-Верхнее давление
    lowerPressure.add(resultOfReadCSV[current][2].toDouble()) //2-Нижнее давление
}

//Статистика
val statsUpperPressure = upperPressure.descriptiveStatistics
println("Верхнее давление (минимум):: ${statsUpperPressure.min}")
println("Верхнее давление (максимум):: ${statsUpperPressure.max}")
println("----------------------------------")
val statsLowerPressure = lowerPressure.descriptiveStatistics
println("Нижнее давление (Минимум):: ${statsLowerPressure.min}")
println("Нижнее давление (Максимум):: ${statsLowerPressure.max}")
println("----------------------------------")
println("Среднее значение верхнего давления = ${statsUpperPressure.mean}")
println("Среднее значение нижнего давления = ${statsLowerPressure.mean}")
val stdDeviation = """
Cтандартное отклонение - это мера разброса данных относительно их среднего значения.
""".trimIndent()
println(stdDeviation)
println("Cтандартнтое отклонение для Верхнего давления:: ${statsUpperPressure.standardDeviation}")
println("Cтандартнтое отклонение для Нижнего давления:: ${statsLowerPressure.standardDeviation}")
println("----------------------------------")
val modaText = """
Мода - это значение, которое встречается в наборе данных чаще всего.
""".trimIndent()
println(modaText)
val modesUpperPressure = upperPressure.mode()  // Возвращает Set (может быть несколько мод)
val modesLowerPressure = lowerPressure.mode()
println("Наиболее частое Верхнее давление (мода):: ${modesUpperPressure.joinToString(" и ")}")
println("Наиболее частое Нижнее давление (мода):: ${modesLowerPressure.joinToString(" и ")}")
println("----------------------------------")
var percentile : Double = 80.0
val procentilUpper = upperPressure.percentile(percentile)
val procentilLower = lowerPressure.percentile(percentile)
println("$percentile% Верхнего давления  находяться ниже $procentilUpper мм рт. ст.")
println("$percentile% Нижнего давления  находяться ниже $procentilLower мм рт. ст.")
println("----------------------------------")
//Тут считаем долю значений выше порога.
fun dolya(davlenie: Int, data: MutableList<Double>, slovo: String){
    //Делаем неизменяемую копию данных
    val dataRadOnly = data.toList()

    val kolichestvoVishe = dataRadOnly.count { it >=  davlenie}
    val procent: Double = (kolichestvoVishe.toDouble()/dataRadOnly.size) * 100
    val formatted = String.format("%.2f", procent)
    println("$slovo давление было на уровне $davlenie или выше в $formatted% измерений")
}
dolya(140, upperPressure, "Верхнее")
dolya(150, upperPressure, "Верхнее")
dolya(90, lowerPressure, "Нижнее")
println("----------------------------------")

//Количество и частота «пиков»
//Автокорреляция
```
