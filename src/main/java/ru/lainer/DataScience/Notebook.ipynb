```kotlin
//CSV файл создал Yandex-ИИ, на основе элекронного письма с давлением.
//Записать сюда промт создания CSV файла в Yandex-ИИ

@file:DependsOn("com.opencsv:opencsv:5.7.1")
@file:DependsOn("org.nield:kotlin-statistics:1.2.1") //от 2018 года

import com.opencsv.CSVReader
import java.io.FileReader
import org.nield.kotlinstatistics.descriptiveStatistics
import org.nield.kotlinstatistics.mode
import org.nield.kotlinstatistics.percentile

val pathToFile: String = "/home/source/srcCode/DSAGitHub004/src/main/java/ru/lainer/DataScience/csv/date001/22.09.2026-omron-full-copy.csv"
val fileReader = FileReader(pathToFile)
val csvReader = CSVReader(fileReader)
val resultOfReadCSV = csvReader.readAll()
csvReader.close()

val upperPressure = mutableListOf<Double>() // Список значений верхнего давления
val lowerPressure = mutableListOf<Double>() // Список значений нижнего давления

for(current in 1 until resultOfReadCSV.size) {
    upperPressure.add(resultOfReadCSV[current][1].toDouble()) //1-Верхнее давление
    lowerPressure.add(resultOfReadCSV[current][2].toDouble()) //2-Нижнее давление
}

//Статистика
println("----------------------------------")
//Верхнее давление, протестил, все правильно.
val statsUpperPressure = upperPressure.descriptiveStatistics
println("Верхнее давление (минимум):: ${statsUpperPressure.min}")
println("Верхнее давление (максимум):: ${statsUpperPressure.max}")
println("----------------------------------")
//Нижнее давление, протестил, все правильно.
val statsLowerPressure = lowerPressure.descriptiveStatistics
println("Нижнее давление (Минимум):: ${statsLowerPressure.min}")
println("Нижнее давление (Максимум):: ${statsLowerPressure.max}")
println("----------------------------------")
//Средне значение (верхнее, нижннее) протестировал:: все правильно
println("Среднее значение верхнего давления = ${statsUpperPressure.mean}")
println("Среднее значение нижнего давления = ${statsLowerPressure.mean}")
println("----------------------------------")
//Стандартное отклонение для верхнее и нижнее давление:: все правильно
val stdDeviation = """
Cтандартное отклонение - это мера разброса данных относительно их среднего значения.
""".trimIndent()
println(stdDeviation)
println("Cтандартнтое отклонение для Верхнего давления:: ${statsUpperPressure.standardDeviation}")
println("Cтандартнтое отклонение для Нижнего давления:: ${statsLowerPressure.standardDeviation}")
println("----------------------------------")
//Мода для верхнего и нижнего давления:: все правильно
val modaText = """
Мода - это значение, которое встречается в наборе данных чаще всего.
""".trimIndent()
println(modaText)
val modesUpperPressure = upperPressure.mode()  // Возвращает Set (может быть несколько мод)
val modesLowerPressure = lowerPressure.mode()
println("Наиболее частое Верхнее давление (мода):: ${modesUpperPressure.joinToString(" и ")}")
println("Наиболее частое Нижнее давление (мода):: ${modesLowerPressure.joinToString(" и ")}")
println("----------------------------------")
//Округленная мода для верхнего и нижнего давления:: все правильно
val modaTextRound = """
Здесь тоже Мода, но на округленных значениях давления.
Округлял в большую сторону.
Например:: 122 до 125, 127 до 130.
""".trimIndent()
println(modaTextRound)
val roundedUpperPressure = upperPressure.map { value ->
    ((value + 4).toInt() / 5) * 5
}
val roundedLowerPressure = lowerPressure.map { value ->
    ((value + 4).toInt() / 5) * 5
}
val modesRoundedUpperPressure = roundedUpperPressure.mode()
val modesRoundedLowerPressure = roundedLowerPressure.mode()
println("Наиболее частое Верхнее давление (округленная мода):: ${modesRoundedUpperPressure.joinToString(" и ")}")
println("Наиболее частое Нижнее давление (окргуленная мода):: ${modesRoundedLowerPressure.joinToString(" и ")}")
println("----------------------------------")
//Процентиль 80 правльно посчитал и правильно интерпретировал для верхнего и нижнего давления.
var percentileValue : Double = 80.0
val procentilUpper = upperPressure.percentile(percentileValue)
val procentilLower = lowerPressure.percentile(percentileValue)
println("$percentileValue% Верхнего давления  находяться ниже или равно $procentilUpper мм рт. ст.")
println("$percentileValue% Нижнего давления  находяться ниже или равно $procentilLower мм рт. ст.")
println("----------------------------------")
//Долю значений выше порога для верхнего и нижнего давления:: посчитал правильно.
//Тут считаем долю значений выше порога.
fun dolya(davlenie: Double, data: MutableList<Double>, slovo: String){
    //Делаем неизменяемую копию данных
    val dataRadOnly = data.toList()

    val kolichestvoVishe = dataRadOnly.count { it >=  davlenie}
    val procent: Double = (kolichestvoVishe.toDouble()/dataRadOnly.size) * 100
    val formatted = String.format("%.2f", procent)
    println("$slovo давление было на уровне $davlenie или выше в $formatted% измерений")
}
dolya(120.0, upperPressure, "Верхнее")
dolya(130.0, upperPressure, "Верхнее")
dolya(140.0, upperPressure, "Верхнее")
dolya(70.0, lowerPressure, "Нижнее")
dolya(80.0, lowerPressure, "Нижнее")
dolya(90.0, lowerPressure, "Нижнее")
println("----------------------------------")

/**
 * Считает выборочную автокорреляцию для заданного лага.
 * Работает с List<Double>.
 */
fun autocorrelation(data: List<Double>, lag: Int): Double {
    require(data.size > lag) { "Для лага $lag нужно хотя бы ${lag + 1} измерений" }

    val n = data.size - lag
    val x = data.dropLast(lag)
    val y = data.drop(lag)

    val meanX = x.average()
    val meanY = y.average()

    var numerator = 0.0
    var denomX = 0.0
    var denomY = 0.0

    for (i in 0 until n) {
        val dx = x[i] - meanX
        val dy = y[i] - meanY
        numerator += dx * dy
        denomX += dx * dx
        denomY += dy * dy
    }

    if (denomX == 0.0 || denomY == 0.0) return 0.0

    return numerator / sqrt(denomX * denomY)
}

/**
 * Возвращает список автокорреляций для лагов от 1 до maxLag.
 */
fun autocorrelations(data: List<Double>, maxLag: Int): List<Double> {
    return (1..maxLag).map { autocorrelation(data, it) }
}

// Считаем автокорреляцию для верхнего и нижнего давления (например, лаги 1–5)
val maxLag = 5
val acfUpper = autocorrelations(upperPressure, maxLag)
val acfLower = autocorrelations(lowerPressure, maxLag)

println("Автокорреляция (ACF) для верхнего давления (лаги 1..$maxLag):")
acfUpper.forEachIndexed { idx, value ->
    println("Лаг ${idx + 1}: ${String.format("%.4f", value)}")
}
println("----------------------------------")
println("Автокорреляция (ACF) для нижнего давления (лаги 1..$maxLag):")
acfLower.forEachIndexed { idx, value ->
    println("Лаг ${idx + 1}: ${String.format("%.4f", value)}")
}
// Добавить анализ результатов автокореляции.

//CSV файл создал Yandex-ИИ, на основе элекронного письма с давлением.
//Записать сюда промт создания CSV файла в Yandex-ИИ

/*
график ACF (коррелограмму) и добавьте на него «полосы значимости» (например, ±2/√n, где n — объём выборки). Значения, выходящие за эти границы, с большей вероятностью статистически значимы, а не случайный шум.
 */
```
