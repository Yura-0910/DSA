```kotlin
println("Исходные данные:: ")
println("---------------------------------------------------")

//Высота потолка
var visotaPotolka : Double = 247.0 //Мои данные:: 247.0
println("Высота потолка = ${visotaPotolka} сантиметров")
println("Высота потолка = ${visotaPotolka/100} метров")
println("---------------------------------------------------")

//Длина и ширина дома
var dlina : Double = 495.0          //С севера на юг, мои данные:: 495.0
var shirina: Double = 425.0 + 177.0 //С востока на запад, мои данные:: 425.0 + 177.0
println("Длина дома (с Севера на Юг) = ${dlina} сантиметров")
println("Длина дома (с Севера на Юг) = ${dlina/100} метров")
println("Ширина дома (с Востока на Запад) = ${shirina} сантиметров")
println("Ширина дома (с Востока на Запад) = ${shirina/100} метров")
println("---------------------------------------------------")

// Минусуем дверь в зал (с кухни)
var dverZalShirina : Double = 115.0 //Мои данные:: 115.0
var dverZalVisota: Double = 201.0   //Мои данные:: 201.0
println("Дверь в зал (с кухни), ширина = ${dverZalShirina} сантиметров")
println("Дверь в зал (с кухни), ширина = ${dverZalShirina/100} метров")
println("Дверь в зал (с кухни), высота = ${dverZalVisota} сантиметров")
println("Дверь в зал (с кухни), высота = ${dverZalVisota/100} метров")
println("---------------------------------------------------")

// Минусуем шесть окон
var oknoShirina: Double = 100.0 //Мои данные:: 100.0
var oknoVisota: Double = 148.0  //Мои данные:: 148.0
println("Ширина окна = ${oknoShirina} сантиметров")
println("Ширина окна = ${oknoShirina/100} метров")
println("Высота окна = ${oknoVisota} сантиметров")
println("Высота окна = ${oknoVisota/100} метров")
println("---------------------------------------------------")

//Считаем площадь стен в зале, без перегородки
var ploshadSten: Double = (dlina + shirina) * visotaPotolka
println("Площадь стен в зале (без учета перегородки) = ${ploshadSten} сантиметров квадратных")
println("Площадь стен в зале (без учета перегородки) = ${ploshadSten * 0.0001} метров квадратных")
println("---------------------------------------------------")

//Минус площадь двери в зал
var minusDverVZal: Double = dverZalShirina * dverZalVisota
println("Площадь двери в зал = ${minusDverVZal} сантиметров квадратных")
println("Площадь двери в зал = ${minusDverVZal  * 0.0001} метров квадратных")
println("---------------------------------------------------")

//Минус площадь шести окон
var minus6Okon : Double = (oknoShirina * oknoVisota) * 6
println("Площадь шести окон = ${minus6Okon} сантиметров квадратных")
println("Площадь шести окон = ${minus6Okon * 0.0001} метров квадратных")
println("---------------------------------------------------")

//Считаем площадь стен в зале (без перегородки), минус площадь двери и минус площадь шести окон.
var itogovayPloshadSten = ploshadSten - minusDverVZal - minus6Okon
println("Считаем площадь стен в зале (без перегородки), минус площадь двери и минус площадь шести окон.")
println("Площадь стен в зале:: общая площадь ${ploshadSten * 0.0001} метров минус площадь двери ${minusDverVZal * 0.0001} метров минус площадь шести окон ${minus6Okon * 0.0001} метров получиться ${itogovayPloshadSten * 0.0001} метров квадратных")
println("---------------------------------------------------")

println("-----------Считаем площадь перегородки-------------")

var dlinaPeregorodka : Double = 495.0 //Мои данные (с Севера на Юг):: 495.0
println("Длина перегородки = ${dlinaPeregorodka} сантиметров")
println("Длина перегородки = ${dlinaPeregorodka/100} метров")
println("---------------------------------------------------")

//Минусуем проход к маленькой кровати
var prohodMalKrovatShirina: Double = 86.0 //Мои данные:: 86.0
var prohodMalKrovatVisota: Double = 187.0 //Мои данные:: 187.0
println("Ширина прохода к маленькой кровати = ${prohodMalKrovatShirina} сантиметров")
println("Ширина прохода к маленькой кровати = ${prohodMalKrovatShirina/100} метров")
println("Высота прохода к маленкой кровати = ${prohodMalKrovatVisota} сантиметров")
println("Высота прохода к маленкой кровати = ${prohodMalKrovatVisota/100} метров")
println("---------------------------------------------------")

// Минусуем печку
var peckaShirina: Double = 101.0 //Мои данные:: 101.0
var pechkaVisota: Double = 247.0 //Мои данные:: 247.0
println("Ширина печки = ${peckaShirina} сантиметров")
println("Ширина печки = ${peckaShirina/100} метров")
println("Высота печки = ${pechkaVisota} сантиметров")
println("Высота печки = ${pechkaVisota/100} метров")
println("---------------------------------------------------")

println("------Считаем площадь перегородки------------------")
var ploshadPeregorodki : Double = dlinaPeregorodka * visotaPotolka
println("Площадь перегородки = ${ploshadPeregorodki} сантиметьров квадратных")
println("Площадь перегородки = ${ploshadPeregorodki * 0.0001} метров квадратных")
println("---------------------------------------------------")

//Считаем площадь прохода к маленькой кровати
var minusProhodMalKrovat : Double = prohodMalKrovatShirina * prohodMalKrovatVisota
println("Площадь прохода к маленькой кровати = ${minusProhodMalKrovat} сантиметров квадратных")
println("Площадь прохода к маленькой кровати = ${minusProhodMalKrovat * 0.0001} метров квадратных")
println("---------------------------------------------------")

//Считаем площадь печки
var minusPeckka : Double = peckaShirina * pechkaVisota
println("Площадь печки = ${minusPeckka} сантиметров квадратных")
println("Площадь печки = ${minusPeckka * 0.0001} метров квадратных")
println("---------------------------------------------------")

println("---------Считаем итоговую площадь перегородки------")
var itogovayPloshadPeregorodki: Double = ploshadPeregorodki - minusProhodMalKrovat - minusPeckka
println("Итоговоая площадь перегородки (с одной стороны) = площадь перегородки ${ploshadPeregorodki * 0.0001} метров квадртаных минус площадь прохода ${minusProhodMalKrovat * 0.0001} метров квадратных минус площадь печки ${minusPeckka * 0.0001} метров квадратных, то получиться, что итоговая площаь перегородки = ${itogovayPloshadPeregorodki * 0.0001} метров квадратных")
println("---------------------------------------------------")

println("---Считаем итоговую площадь зала:: площадь стен + площадь перегородки (с одной строны)---")
var itogovayPloshadSm = itogovayPloshadSten + itogovayPloshadPeregorodki
var itogovayPloshadMetri: Double = itogovayPloshadSm * 0.0001
println("Итоговоая площадь -это площадь стен зала + площадь перегородки с одной сторионы = итоговая площадь стен зала = ${itogovayPloshadSten * 0.0001} метров квадратных плюс итоговая площадь перегородки в зале (с одной стороны) = ${itogovayPloshadPeregorodki * 0.0001} метров квадратных, получиться в итоге:: ${itogovayPloshadSm *0.0001 } метров квадратных")
println("---------------------------------------------------")

var cenaZaMetrKvadratniy: Double = 300.0 //руб за метр квадратный
var itogovayaCena: Double = itogovayPloshadMetri * cenaZaMetrKvadratniy
println("Если цена за кв.метр = ${cenaZaMetrKvadratniy} руб., то поклеить стоит = ${itogovayaCena} руб.")
```
